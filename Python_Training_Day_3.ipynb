# Writing data to .CSV file

```python
import csv
with open("emp.csv", "w", newline="") as f:
    w = csv.writer(f)
    w.writerow(["ENO", "ENAME", "ESALARY", "EADDR"])
    n = int(input("Enter number of employees : "))
    for i in range(3):
        eno = input("Enter employee no : ")
        ename = input("Enter employee name : ")
        esal = input("Enter employee salary : ")
        eaddr = input("Enter employee address : ")
        w.writerow([eno, ename, esal, eaddr])

print("Data inserted successfully!")
```

# Reading data from .CSV file

```python
import csv
f=open("emp.csv","r")
r=csv.reader(f)
data=list(r)
print(data)
for line in data:
    for word in line:
        print(word,"\t",end='')
    print()
```

#NumPy

```python
import numpy as np
students = np.array([
    [85, 80, 70],  # Student 1
    [80, 55, 70],  # Student 2
    [85, 92, 88],  # Student 3
    [80, 90, 45],  # Student 4
    [72, 68, 75]   # Student 5
])
students
```

```python
avg_marks = np.mean(students, axis=1)
print("Average Nakrs :", avg_marks)
```

```python
topper = np.argmax(avg_marks)
print("Topper Students :", topper)
```

```python
failed_student = np.where(avg_marks < 70)
print("Failed Students :", failed_student[0]+1)
```

#Performance report

```python
for i, avg in enumerate(avg_marks):
  status = "Pass" if avg >= 70 else "Fail"
  print(f"Student {i+1}: Average Marks = {avg:.2f} | {status}")
```

```python
print(f"{'Student':<10}{'Average Marks':<15}{'Status'}")

for i, avg in enumerate(avg_marks):
    status = "Pass" if avg >= 70 else "Fail"
    print(f"{i+1:<10}{avg:<15.2f}{status}")
```

#Grouping and Aggregation

#Sorting data

```python
import pandas as pd
data = {
    "Name": ["John Doe", "Nakul", "Anshul"],
    "Age": [30, 20, 25],
    "City": ["New York", "Los Angeles", "New York"]
}
df = pd.DataFrame(data)
df.groupby('City').agg({'Age': ['mean', 'max']})
```

```python
df.sort_values(by='Age', ascending=False)
```

#value count

```python
df['City'].value_counts()
pd.crosstab(df['City'], df['Age'])
```

#Bar Chart

```python
months = ['Jan','Feb','Mar','Apr','May']
sales = [100, 150, 120, 180, 200]
plt.bar(months, sales, color='#35063e')
plt.title('Sales Analysis')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.show()
```

```python
plt.barh(months, sales, color='#35063e')
plt.title('Monthly Sales (Horizontal)')
plt.xlabel('sales')
plt.ylabel('months')
plt.show()
```

#Scatter plot (comparing height vs weight)

```python
height = [150, 160, 170, 180, 190]
weight = [50, 60, 70, 80, 90]
plt.scatter(height, weight, color='#033E3E')
plt.title('Height vs Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

#Pie chart market share

```python
brands = ['Brand A', 'Brand B', 'Brand C', 'Brand D']
share = [20, 30, 40, 10]
plt.pie(share, labels=brands, autopct='%1.1f%%', startangle=90)
plt.title('Market Share')
plt.axis('equal')
plt.show()
```

#Multiple plot in one figure

```python
x = [1, 2, 3, 4, 5]
y1 = [10, 20, 30, 40, 50]
y2 = [5, 15, 30, 45, 75]
plt.plot(x, y1, label='square')
plt.plot(x, y2, label='Linear')
plt.legend()
plt.title('Multiple Plots')
plt.show()
```

#Subplots

```python
fig, axs = plt.subplots(2, 2)

axs[0, 0].plot(x, y1)
axs[0, 0].set_title('Square')

axs[0, 1].bar(months, sales)
axs[0, 1].set_title('Sales')

axs[1, 0].scatter(height, weight)
axs[1, 0].set_title('Scatter')

axs[1, 1].pie(share, labels=brands, autopct='%1.1f%%')
axs[1, 1].set_title('Pie')

plt.tight_layout()
plt.show()
```
